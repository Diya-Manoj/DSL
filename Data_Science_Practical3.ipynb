# Part 1: Summary Statistics Grouped by Categorical Variable
using the tips dataset from Seaborn, where we'll calculate summary stats for total_bill grouped by day.

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

##  Load the tips dataset

```python
 df = sns.load_dataset('tips')
```

## Display sample of the dataset

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

## Group by a categorical variable ('day') and summarize a numeric one ('total_bill')

```python
summary = df.groupby('day')['total_bill'].agg(['mean', 'median', 'min', 'max', 'std'])
print("\nSummary statistics for 'total_bill' grouped by 'day':")
summary
```

##  Create list of values for each category

```python
grouped_values = df.groupby('day')['total_bill'].apply(list).to_dict()
print("\nNumeric value lists for each day:")
print(grouped_values)
```

```python

sns.boxplot(x='day', y='total_bill', data=df)
```

# Part 2: Summary Statistics for Iris Dataset (Grouped by Species)

## Load the iris Dataset

```python
from sklearn.datasets import load_iris
iris = load_iris(as_frame=True)
df_iris = iris.frame
df_iris
```

## Combine features and target

```python
df_iris['species'] = iris.target
df_iris['species'] = df_iris['species'].map({0:'setosa', 1:'versicolor', 2:'virginica'})
```

```python
df_iris
```

## Filter Species and compute stats

```python
for species in ['setosa', 'versicolor', 'virginica']:
    print(f"\n--- Statistics for {species} ---")
    subset = df_iris[df_iris['species'] == species].drop('species', axis=1)
    desc = subset.describe(percentiles=[.25, .5, .75])
    print(desc)
```

```python
for species in df_iris['species'].unique():
    subset = df_iris[df_iris['species'] == species]
    plt.hist(subset['petal length (cm)'], bins=15, alpha=0.6, label=species)

'''for species in df_iris['species'].unique():      	Loops through each unique species (Setosa, Versicolor, Virginica) dynamically.
   subset = df_iris[df_iris['species'] == species]  	Filters the dataset for only the rows of the current species.
   plt.hist(...)                                    	Plots a histogram of the petal length (cm) for that species.
   bins=15                                          	Breaks the x-axis into 15 bins (intervals) for better resolution of the distribution.
   alpha=0.6                                        	Sets the transparency level (60%) to allow overlapping plots to be visible.
'''    
plt.title('Distribution of Petal Length by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
