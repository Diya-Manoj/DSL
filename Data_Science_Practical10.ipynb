## Iris Dataset Visualization & Analysis

### Step 1: Load the Dataset

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Iris dataset
iris = sns.load_dataset('iris')

# Display first few rows
print(iris.head())
```

### Step 2. Histograms for Each Feature

Inference:
* Petal length and petal width show clear separation between species (bimodal/multimodal).
* Sepal length is more evenly spread.
* Sepal width is slightly left-skewed.

```python
iris.hist(edgecolor='black', figsize=(10, 8))
plt.suptitle('Histograms of Iris Dataset Features')
plt.tight_layout()
plt.show()
```

### Step 3. Boxplots for Each Feature

 Inference:

* Petal width and length show distinct variation and separation among species.
* Outliers are visible especially in sepal width and sepal length.
* Helps in identifying variability and skewness.

```python
for column in iris.columns[:-1]:  # Exclude 'species'
    sns.boxplot(y=iris[column])
    plt.title(f'Boxplot of {column}')
    plt.grid(True)
    plt.show()
```

###  Key Observations:
* Petal features are the most important for distinguishing species — petal width for setosa is very small compared to others.
* Sepal width contains visible outliers.
* Species setosa tends to have smaller petal dimensions, creating clearer separation.

Sepal Length and Petal Length: Fairly normally distributed.

Sepal Width: Slightly skewed distribution, potential outliers.

Petal Width: Shows a clear peak around one value (indicative of class separation).

Outliers: Some present in sepal_width.
