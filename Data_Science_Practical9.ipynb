##  1. Box Plot of Age vs Gender with Survival Info

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load Titanic dataset
titanic = sns.load_dataset('titanic')
titanic
```

```python
# Plotting the boxplot
sns.boxplot(x='sex', y='age', hue='survived', data=titanic)
plt.title('Age Distribution by Gender and Survival Status')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.legend(title='Survived')
plt.grid(True)
plt.show()
```

## Observations:
#### 1. Younger females had a higher survival rate.

The median age of surviving females is lower than that of non-surviving females, suggesting younger women were more likely to be rescued.

#### 2. Males had a lower survival rate overall.

The box for surviving males is narrower and lower, showing fewer males survived, and those who did were mostly younger.

#### 3. More outliers among older males.

The plot shows several male passengers who were older but did not survive, visible as points above the whiskers.

#### 4. Age range was wider for non-survivors.

Both males and females who didn't survive show a wider spread of ages, from infants to elderly passengers.

#### 5. Survival was not strictly dependent on age, but gender and age together had a strong influence.
