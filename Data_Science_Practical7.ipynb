## Sample Text and Preprocessing

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
import string
```

## Download required NLTK resources

```python
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('stopwords')
nltk.download('wordnet')
```

## Sample document

```python
text = "The quick brown foxes were jumping over the lazy dogs near the riverbank."
```

## 1. Tokenization

```python
tokens = word_tokenize(text)
print("Tokens:", tokens)
```

## 2. POS Tagging

```python
pos_tags = pos_tag(tokens)
print("POS Tags:", pos_tags)
```

## 3. Stopwords Removal

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word.lower() not in stop_words and word not in string.punctuation]
print("Filtered Tokens (Stopwords removed):", filtered_tokens)
```

## 4. Stemming

```python
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
print("Stemmed Tokens:", stemmed_tokens)
```

## 5. Lemmatization

```python
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word.lower()) for word in filtered_tokens]
print("Lemmatized Tokens:", lemmatized_tokens)
```

## Task 2: TF-IDF Representation

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample documents
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "Never jump over the lazy dog quickly.",
    "Foxes are quicker than dogs near the river."
]
```

## Create TF-IDF Vectorizer

```python
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)
```

## Convert to DataFrame for readability

```python
import pandas as pd
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())
print("\nTF-IDF Representation:")
df_tfidf
```
